```python
import pandas as pd

ccc_full = pd.read_csv(
    "/content/ccc_compiled_20212024.csv",
    encoding="latin1",
    low_memory=False
)

# Convert dates
ccc_full["date"] = pd.to_datetime(
    ccc_full["date"], errors="coerce"
)

print("Total records:", len(ccc_full))
print("Date range:", ccc_full["date"].min(), "to", ccc_full["date"].max())
print("Number of states:", ccc_full["state"].nunique())

print("\nEvents by year:")
print(ccc_full["date"].dt.year.value_counts().sort_index())

print("\nTop 15 states by event count:")
print(ccc_full["state"].value_counts().head(15))

print("\nEvent types:")
print(ccc_full["type"].value_counts().head(15))
```

```python
# Check missing values in important columns
important_cols = [
    "date", "state", "type", "size_mean",
    "arrests", "injuries_crowd", "property_damage"
]

print("Missing values:")
print(ccc_full[important_cols].isnull().sum())

print("\nSample records:")
display(ccc_full[important_cols].head(10))

print("\nUnique state codes:")
print(sorted(ccc_full["state"].dropna().unique()))
```

```python
import pandas as pd

# Remove records without a state
ccc_clean = ccc_full.dropna(subset=["state"]).copy()

# Ensure date format
ccc_clean["date"] = pd.to_datetime(ccc_clean["date"])

# Count events per state per day
daily_events = (
    ccc_clean.groupby(["date", "state"])
    .size()
    .reset_index(name="event_count")
)

print("Daily state-level records:", len(daily_events))

print("\nSample aggregated data:")
display(daily_events.head(15))

print("\nMaximum events in a single state-day:",
      daily_events["event_count"].max())
```

```python
import pandas as pd

all_dates = pd.date_range(
    daily_events["date"].min(),
    daily_events["date"].max(),
    freq="D"
)

all_states = sorted(daily_events["state"].unique())

full_index = pd.MultiIndex.from_product(
    [all_dates, all_states],
    names=["date", "state"]
)

daily_full = daily_events.set_index(["date", "state"]).reindex(
    full_index, fill_value=0
).reset_index()

daily_full["event_count"] = daily_full["event_count"].astype(int)

print("Total state-day combinations:", len(daily_full))
print("Total dates:", daily_full["date"].nunique())
print("Total states:", daily_full["state"].nunique())

print("\nSample:")
display(daily_full.head(15))

print("\nZero-event days:", (daily_full["event_count"] == 0).sum())
```

```python
# Sort data chronologically for each state
daily_full = daily_full.sort_values(["state", "date"]).copy()

# Lag features: previous event counts
for lag in [1, 3, 7, 14]:
    daily_full[f"lag_{lag}"] = (
        daily_full.groupby("state")["event_count"].shift(lag)
    )

# Rolling features using only past data
daily_full["rolling_7_mean"] = (
    daily_full.groupby("state")["event_count"]
    .transform(lambda x: x.shift(1).rolling(7).mean())
)

daily_full["rolling_7_sum"] = (
    daily_full.groupby("state")["event_count"]
    .transform(lambda x: x.shift(1).rolling(7).sum())
)

daily_full["rolling_14_mean"] = (
    daily_full.groupby("state")["event_count"]
    .transform(lambda x: x.shift(1).rolling(14).mean())
)

print("Feature creation complete.")

print("\nSample features:")
display(daily_full.head(20))
```

```python
import numpy as np

# Predict whether any event will occur in the next 1, 3, or 7 days
for horizon in [1, 3, 7]:
    future_counts = pd.concat(
        [
            daily_full.groupby("state")["event_count"].shift(-day)
            for day in range(1, horizon + 1)
        ],
        axis=1
    )

    daily_full[f"target_{horizon}d"] = (
        future_counts.max(axis=1)
    )

    # Rows without a complete future horizon are excluded
    daily_full.loc[
        future_counts.isna().any(axis=1),
        f"target_{horizon}d"
    ] = np.nan

    # Convert event counts to binary labels
    daily_full[f"target_{horizon}d"] = (
        daily_full[f"target_{horizon}d"].apply(
            lambda x: 1 if x > 0 else 0
        )
        if daily_full[f"target_{horizon}d"].notna().all()
        else daily_full[f"target_{horizon}d"].map(
            lambda x: 1 if x > 0 else (0 if x == 0 else np.nan)
        )
    )

print("Forecast targets created.")

print("\nTarget distribution:")
for horizon in [1, 3, 7]:
    print(f"\nNext {horizon} days:")
    print(daily_full[f"target_{horizon}d"].value_counts(dropna=False))

display(
    daily_full[[
        "date", "state", "event_count",
        "target_1d", "target_3d", "target_7d"
    ]].head(20)
)
```

```python
# Feature columns
feature_cols = [
    "lag_1", "lag_3", "lag_7", "lag_14",
    "rolling_7_mean", "rolling_7_sum",
    "rolling_14_mean"
]

# Remove rows without sufficient historical features
model_data = daily_full.dropna(subset=feature_cols).copy()

# Chronological splits
train_data = model_data[
    model_data["date"] < "2023-01-01"
].copy()

val_data = model_data[
    (model_data["date"] >= "2023-01-01") &
    (model_data["date"] < "2024-01-01")
].copy()

test_data = model_data[
    model_data["date"] >= "2024-01-01"
].copy()

print("Training records:", len(train_data))
print("Validation records:", len(val_data))
print("Testing records:", len(test_data))

print("\nTraining date range:",
      train_data["date"].min(), "to", train_data["date"].max())

print("\nValidation date range:",
      val_data["date"].min(), "to", val_data["date"].max())

print("\nTesting date range:",
      test_data["date"].min(), "to", test_data["date"].max())
```

```python
# Prepare clean splits for each forecasting horizon

splits = {}

cutoffs = {
    "train": "2022-12-31",
    "val": "2023-12-31",
    "test": "2024-12-31"
}

for horizon in [1, 3, 7]:

    target_col = f"target_{horizon}d"

    splits[horizon] = {}

    for split_name, cutoff in cutoffs.items():

        split_start = {
            "train": "2021-01-01",
            "val": "2023-01-01",
            "test": "2024-01-01"
        }[split_name]

        # Keep only dates whose full forecast horizon
        # stays within that split's time boundary
        end_date = pd.Timestamp(cutoff) - pd.Timedelta(days=horizon)

        subset = model_data[
            (model_data["date"] >= split_start) &
            (model_data["date"] <= end_date)
        ].copy()

        subset = subset.dropna(subset=[target_col])

        X = subset[feature_cols]
        y = subset[target_col].astype(int)

        splits[horizon][split_name] = (X, y)

        print(
            f"{horizon}-day | {split_name}: "
            f"{len(X)} records | "
            f"Positive: {y.sum()} | "
            f"Negative: {(y == 0).sum()}"
        )

print("\nAll forecasting datasets prepared.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, f1_score, precision_score, recall_score
from xgboost import XGBClassifier

results = []

for horizon in [1, 3, 7]:

    X_train, y_train = splits[horizon]["train"]
    X_val, y_val = splits[horizon]["val"]

    # -----------------------------
    # 1. Persistence baseline
    # -----------------------------
    val_rows = model_data[
        (model_data["date"] >= "2023-01-01") &
        (model_data["date"] <= pd.Timestamp("2023-12-31") - pd.Timedelta(days=horizon))
    ].dropna(subset=[f"target_{horizon}d"])

    persistence_pred = (val_rows["rolling_7_sum"] > 0).astype(int)
    persistence_true = val_rows[f"target_{horizon}d"].astype(int)

    results.append({
        "Horizon": f"{horizon}-day",
        "Model": "Persistence",
        "PR-AUC": average_precision_score(persistence_true, persistence_pred),
        "F1": f1_score(persistence_true, persistence_pred, zero_division=0),
        "Precision": precision_score(persistence_true, persistence_pred, zero_division=0),
        "Recall": recall_score(persistence_true, persistence_pred, zero_division=0)
    })

    # -----------------------------
    # 2. Logistic Regression
    # -----------------------------
    lr = Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ))
    ])

    lr.fit(X_train, y_train)
    lr_prob = lr.predict_proba(X_val)[:, 1]
    lr_pred = (lr_prob >= 0.5).astype(int)

    results.append({
        "Horizon": f"{horizon}-day",
        "Model": "Logistic Regression",
        "PR-AUC": average_precision_score(y_val, lr_prob),
        "F1": f1_score(y_val, lr_pred, zero_division=0),
        "Precision": precision_score(y_val, lr_pred, zero_division=0),
        "Recall": recall_score(y_val, lr_pred, zero_division=0)
    })

    # -----------------------------
    # 3. XGBoost
    # -----------------------------
    scale_pos_weight = (y_train == 0).sum() / max((y_train == 1).sum(), 1)

    xgb = XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    xgb.fit(X_train, y_train)

    xgb_prob = xgb.predict_proba(X_val)[:, 1]
    xgb_pred = (xgb_prob >= 0.5).astype(int)

    results.append({
        "Horizon": f"{horizon}-day",
        "Model": "XGBoost",
        "PR-AUC": average_precision_score(y_val, xgb_prob),
        "F1": f1_score(y_val, xgb_pred, zero_division=0),
        "Precision": precision_score(y_val, xgb_pred, zero_division=0),
        "Recall": recall_score(y_val, xgb_pred, zero_division=0)
    })

# Display comparison
results_df = pd.DataFrame(results)

print("Validation baseline comparison:")
display(results_df.round(4))
```

```python
import numpy as np
from sklearn.metrics import f1_score

threshold_results = []

thresholds = np.arange(0.10, 0.91, 0.05)

for horizon in [1, 3, 7]:

    X_val, y_val = splits[horizon]["val"]

    # Retrieve the XGBoost model trained earlier
    xgb_model = None

    # Refit using the same training data
    X_train, y_train = splits[horizon]["train"]

    scale_pos_weight = (y_train == 0).sum() / max((y_train == 1).sum(), 1)

    xgb_model = XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    xgb_model.fit(X_train, y_train)

    probabilities = xgb_model.predict_proba(X_val)[:, 1]

    best_f1 = 0
    best_threshold = 0.5

    for threshold in thresholds:
        predictions = (probabilities >= threshold).astype(int)
        score = f1_score(y_val, predictions, zero_division=0)

        if score > best_f1:
            best_f1 = score
            best_threshold = threshold

    threshold_results.append({
        "Horizon": f"{horizon}-day",
        "Best Threshold": best_threshold,
        "Best Validation F1": best_f1
    })

threshold_df = pd.DataFrame(threshold_results)

display(threshold_df.round(4))
```

```python
from sklearn.metrics import (
    f1_score, precision_score, recall_score,
    average_precision_score, accuracy_score,
    confusion_matrix, classification_report
)
import pandas as pd
import numpy as np

selected_thresholds = {
    1: 0.35,
    3: 0.25,
    7: 0.10
}

final_results = []

for horizon in [1, 3, 7]:

    X_train, y_train = splits[horizon]["train"]
    X_test, y_test = splits[horizon]["test"]

    threshold = selected_thresholds[horizon]

    scale_pos_weight = (y_train == 0).sum() / max((y_train == 1).sum(), 1)

    model = XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    probabilities = model.predict_proba(X_test)[:, 1]
    predictions = (probabilities >= threshold).astype(int)

    final_results.append({
        "Horizon": f"{horizon}-day",
        "Threshold": threshold,
        "PR-AUC": average_precision_score(y_test, probabilities),
        "F1": f1_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "Accuracy": accuracy_score(y_test, predictions)
    })

    print(f"\n{'='*50}")
    print(f"{horizon}-DAY FORECAST")
    print(f"{'='*50}")
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, predictions))
    print("\nClassification Report:")
    print(classification_report(y_test, predictions, zero_division=0))

final_results_df = pd.DataFrame(final_results)

print("\nFINAL TEST RESULTS (2024)")
display(final_results_df.round(4))
```

```python
from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score, average_precision_score, confusion_matrix
import pandas as pd

selected_thresholds = {1: 0.35, 3: 0.25, 7: 0.10}

refined_results = []

for horizon in [1, 3, 7]:

    X_train, y_train = splits[horizon]["train"]
    X_val, y_val = splits[horizon]["val"]
    X_test, y_test = splits[horizon]["test"]

    # Combine training and validation data
    X_combined = pd.concat([X_train, X_val])
    y_combined = pd.concat([y_train, y_val])

    scale_pos_weight = (
        (y_combined == 0).sum() / max((y_combined == 1).sum(), 1)
    )

    model = XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_combined, y_combined)

    threshold = selected_thresholds[horizon]

    probabilities = model.predict_proba(X_test)[:, 1]
    predictions = (probabilities >= threshold).astype(int)

    refined_results.append({
        "Horizon": f"{horizon}-day",
        "Threshold": threshold,
        "PR-AUC": average_precision_score(y_test, probabilities),
        "F1": f1_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "Accuracy": accuracy_score(y_test, predictions)
    })

    print(f"\n{'='*50}")
    print(f"{horizon}-DAY FORECAST (TRAIN + VALIDATION)")
    print(f"{'='*50}")
    print(confusion_matrix(y_test, predictions))

refined_results_df = pd.DataFrame(refined_results)

print("\nFINAL REFINED TEST RESULTS")
display(refined_results_df.round(4))
```

```python
from sklearn.metrics import average_precision_score, f1_score
import pandas as pd

feature_groups = {
    "Full model": feature_cols,
    "Lag-only": ["lag_1", "lag_3", "lag_7", "lag_14"],
    "Rolling-only": ["rolling_7_mean", "rolling_7_sum", "rolling_14_mean"]
}

ablation_results = []

for horizon in [1, 3, 7]:

    X_train, y_train = splits[horizon]["train"]
    X_val, y_val = splits[horizon]["val"]

    threshold = selected_thresholds[horizon]

    for group_name, selected_features in feature_groups.items():

        X_train_subset = X_train[selected_features]
        X_val_subset = X_val[selected_features]

        scale_pos_weight = (
            (y_train == 0).sum() /
            max((y_train == 1).sum(), 1)
        )

        model = XGBClassifier(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            scale_pos_weight=scale_pos_weight,
            eval_metric="logloss",
            random_state=42,
            n_jobs=-1
        )

        model.fit(X_train_subset, y_train)

        probabilities = model.predict_proba(X_val_subset)[:, 1]
        predictions = (probabilities >= threshold).astype(int)

        ablation_results.append({
            "Horizon": f"{horizon}-day",
            "Feature Set": group_name,
            "PR-AUC": average_precision_score(y_val, probabilities),
            "F1": f1_score(y_val, predictions),
        })

ablation_df = pd.DataFrame(ablation_results)

display(
    ablation_df.pivot(
        index="Feature Set",
        columns="Horizon",
        values="PR-AUC"
    ).round(4)
)

display(
    ablation_df.pivot(
        index="Feature Set",
        columns="Horizon",
        values="F1"
    ).round(4)
)
```

```python
import os
import json
import zipfile
import pandas as pd
from xgboost import XGBClassifier
from sklearn.metrics import (
    f1_score, precision_score, recall_score,
    accuracy_score, average_precision_score
)

# Create artifact folder
os.makedirs("civil_unrest_models", exist_ok=True)

selected_thresholds = {1: 0.35, 3: 0.25, 7: 0.10}

metadata = {
    "project": "Civil Unrest Event Forecasting",
    "dataset": "CCC compiled 2021-2024",
    "features": feature_cols,
    "thresholds": selected_thresholds,
    "forecast_horizons": [1, 3, 7],
    "seed": 42
}

final_saved_results = []

for horizon in [1, 3, 7]:

    X_train, y_train = splits[horizon]["train"]
    X_val, y_val = splits[horizon]["val"]
    X_test, y_test = splits[horizon]["test"]

    # Combine train and validation
    X_combined = pd.concat([X_train, X_val])
    y_combined = pd.concat([y_train, y_val])

    scale_pos_weight = (
        (y_combined == 0).sum() /
        max((y_combined == 1).sum(), 1)
    )

    model = XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_combined, y_combined)

    # Save model
    model_path = f"civil_unrest_models/xgboost_{horizon}day.json"
    model.save_model(model_path)

    # Evaluate on 2024 test set
    probabilities = model.predict_proba(X_test)[:, 1]
    predictions = (probabilities >= selected_thresholds[horizon]).astype(int)

    final_saved_results.append({
        "Horizon": f"{horizon}-day",
        "Threshold": selected_thresholds[horizon],
        "PR-AUC": average_precision_score(y_test, probabilities),
        "F1": f1_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "Accuracy": accuracy_score(y_test, predictions)
    })

# Save metadata
with open("civil_unrest_models/metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

# Save final results
saved_results_df = pd.DataFrame(final_saved_results)
saved_results_df.to_csv(
    "civil_unrest_models/final_results.csv", index=False
)

# Create ZIP
zip_path = "civil_unrest_models.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for root, dirs, files in os.walk("civil_unrest_models"):
        for file in files:
            path = os.path.join(root, file)
            z.write(path, path)

print("MODELS SAVED SUCCESSFULLY")
display(saved_results_df.round(4))

from google.colab import files
files.download(zip_path)
```
